## 4. リポジトリ収録の結果と照合

リポジトリに収録されている結果ファイルと、いま実行した結果が一致するかを確かめます。

```python
import pathlib

for name in ["結果（国別集計）.txt", "結果（rule確認）.txt"]:
    saved = pathlib.Path(name).read_text(encoding="utf-8").replace("\r\n", "\n")
    now = pathlib.Path("output", name).read_text(encoding="utf-8").replace("\r\n", "\n")
    print(name, "一致" if saved == now else "差分あり")
```
